## Question

There are three piles containing 3, 4, and 5 stones. Two players alternate turns. On each turn, a player chooses one pile and removes at least one stone from it. The player who takes the last stone wins.

If you move first, can you force a win against a perfect opponent? What move should you make?

## Solution

Compute the nim-sum (bitwise XOR) of the pile sizes:

$3\;\text{XOR}\;4\;\text{XOR}\;5=2.$

A nonzero nim-sum is a winning position: make a move that changes it to zero. Reducing the first pile from 3 stones to 1 gives $(1,4,5)$, whose nim-sum is zero. Thus, **remove 2 stones from the first pile**. With perfect play, this leaves the opponent a losing position.

## Python


```python
piles = (3, 4, 5)

def nim_sum(position):
    result = 0
    for pile in position:
        result ^= pile
    return result

winning_moves = []
for pile_index, pile_size in enumerate(piles):
    for new_size in range(pile_size):
        next_position = list(piles)
        next_position[pile_index] = new_size
        if nim_sum(next_position) == 0:
            winning_moves.append((pile_index + 1, pile_size - new_size, tuple(next_position)))

print("Starting piles:", piles)
print("Starting XOR (nim-sum):", nim_sum(piles))
print("Moves that leave XOR zero:")
for pile_number, removed, result in winning_moves:
    print(f"  Remove {removed} from pile {pile_number}: {result}")
```

## Python interpretation

The code checks every legal move and finds that removing two stones from pile 1 is the move that leaves XOR zero. This confirms the winning move predicted by the nim-sum strategy.
