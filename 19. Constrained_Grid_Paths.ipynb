## Question

On a 6-by-6 grid, travel from the lower-left corner $(0,0)$ to the upper-right corner $(6,6)$. Each move must go one unit right or one unit up.

There is one extra rule: at every point along the route, you must stay on or below the diagonal $y=x$.

How many valid routes are there?

## Solution

Let $P(x,y)$ be the number of valid paths to $(x,y)$. Points above the diagonal have zero valid paths. On or below it, each path arrives from the left or from below, so

$P(x,y)=P(x-1,y)+P(x,y-1),$

with $P(0,0)=1$ and forbidden points assigned zero. These counts are the Catalan numbers:

$C_n=\frac{1}{n+1}\binom{2n}{n}.$

For $n=6$, $C_6=\frac{1}{7}\binom{12}{6}=\mathbf{132}$ valid routes.

## Python


```python
size = 6
paths = [[0] * (size + 1) for _ in range(size + 1)]
paths[0][0] = 1

for x in range(size + 1):
    for y in range(size + 1):
        if y > x or (x == 0 and y == 0):
            continue

        from_left = paths[x - 1][y] if x > 0 else 0
        from_below = paths[x][y - 1] if y > 0 else 0
        paths[x][y] += from_left + from_below

print(f"Valid routes on a {size} by {size} grid:", paths[size][size])
print("Catalan-number check:", __import__("math").comb(2 * size, size) // (size + 1))
print("\nPath counts at each grid point (x across, y up):")
for y in reversed(range(size + 1)):
    print(" ".join(f"{paths[x][y]:4}" if y <= x else "   ." for x in range(size + 1)))
```

## Python interpretation

The table fills each grid point from its two allowed predecessor points and sets positions above the diagonal to zero. It gives 132 routes, matching the Catalan-number formula; without the restriction there are $\binom{12}{6}=924$ routes.
