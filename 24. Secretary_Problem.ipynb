## Question

Ten applicants interview in a uniformly random order. You can compare each applicant with everyone you've seen so far, but once you reject someone, you cannot go back. You must accept or reject each applicant immediately, and you want to hire the single best applicant.

A simple strategy is: reject the first $r$ applicants to learn the quality range, then hire the first later applicant who is better than everyone in that initial group.

Which $r$ gives this strategy the best chance of hiring the best applicant?

## Solution

If the best applicant is at position $j>r$, the strategy succeeds exactly when the best among the first $j-1$ applicants is in the initial sample of $r$. This occurs with probability $r/(j-1)$. The best applicant is equally likely to occupy each position, so

$P_r=\frac1{10}\sum_{j=r+1}^{10}\frac{r}{j-1}.$

Evaluating this for each possible sample size gives the largest success probability at $r=3$:

$P_3=\frac{3349}{8400}\approx\mathbf{39.869\%}.$ 

So reject the first three applicants, then hire the first later applicant who is better than everyone in that initial group; if none qualifies, take the last applicant.

## Python


```python
from fractions import Fraction

n = 10
success_probabilities = {}

for sample_size in range(n):
    # The best applicant must appear after the sample. If they appear
    # at position j (1-based), they are selected if the best of the
    # previous j-1 applicants was in the sample.
    probability = (
        sum(
            (Fraction(1, n) * Fraction(sample_size, position - 1))
            for position in range(sample_size + 1, n + 1)
        )
        if sample_size > 0 else Fraction(0)
    )
    success_probabilities[sample_size] = probability

best_sample_size = max(success_probabilities, key=success_probabilities.get)
print("Applicants observed before selecting:")
for sample_size, probability in success_probabilities.items():
    print(f"  {sample_size:2}: {float(probability):.3%}")
print(f"\nBest sample size: {best_sample_size}")
print(f"Best success probability: {success_probabilities[best_sample_size]} = "
      f"{float(success_probabilities[best_sample_size]):.3%}")
```

## Python interpretation

The code evaluates the success probability exactly as a fraction for every sample size from 0 to 9. It selects a sample of three, with success probability about **39.869%**, the highest among the strategies considered.
