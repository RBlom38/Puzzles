## Question

Assume every day of a 365-day year is equally likely to be someone's birthday, and ignore leap years. What is the smallest number of people needed for the probability that at least two share a birthday to exceed 50%?


## Solution

Calculate the complement of the probability that no two people share a birthday.

The first person can have any birthday.    
The second must avoid that one day, so the chance is $364/365$.    
The third must avoid two used days, so the chance is $363/365$.    
Continue this pattern for each person.    

$P(\text{at least one shared birthday})=1-\frac{365\cdot364\cdot363\cdots(365-n+1)}{365^n}.$



## Python calculation


```python
DAYS_IN_YEAR = 365

def shared_birthday_probability(number_of_people):
    probability_all_different = 1.0
    for person_index in range(number_of_people):
        probability_all_different *= (DAYS_IN_YEAR - person_index) / DAYS_IN_YEAR
    return 1 - probability_all_different

number_of_people = 1
while shared_birthday_probability(number_of_people) <= 0.5:
    number_of_people += 1

print(f"First group size above 50%: {number_of_people}")
print(f"At {number_of_people - 1} people: {shared_birthday_probability(number_of_people - 1):.1%}")
print(f"At {number_of_people} people: {shared_birthday_probability(number_of_people):.1%}")
```
