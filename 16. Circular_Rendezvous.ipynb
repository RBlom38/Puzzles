## Question

A circular track has 24 equally spaced marks. Three runners start at marks 16, 8, and 0. At each time step, they move clockwise by 1, 2, and 3 marks, respectively. They never stop.

Will all three ever land on the same mark at the same time? If so, when? Check every time in one full cycle, then look for the repeating structure.

## Solution

After $t$ steps, the runners' positions modulo 24 are $16+t$, $8+2t$, and $3t$. For all three to meet, the first two positions must agree:

$16+t\equiv 8+2t\pmod{24},$

which requires $t\equiv 8\pmod{24}$. At $t=8$, all three positions are 0, so they meet at mark 0. Their position pattern repeats after

$\operatorname{lcm}(24,12,8)=24\text{ steps},$

so the first meeting is at step 8, with later meetings every 24 steps.

## Python


```python
track_size = 24
starts = [16, 8, 0]
speeds = [1, 2, 3]

# Every runner returns to their starting mark after a whole number of steps. 
# the combined pattern repeats after the least common multiple.
from math import gcd, lcm
period = lcm(*(track_size // gcd(track_size, speed) for speed in speeds))

positions_by_time = []
meetings = []
for time in range(period + 1):
    positions = [(start + speed * time) % track_size
                 for start, speed in zip(starts, speeds)]
    positions_by_time.append(positions)
    if len(set(positions)) == 1:
        meetings.append((time, positions[0]))

print("Pattern period:", period, "steps")
print("Simultaneous meetings in one period:", meetings)
print("First meeting:", meetings[0][0], "steps at mark", meetings[0][1])
```

## Python interpretation

The code checks one full 24-step period and finds the first meeting at step 8, mark 0. Since the pattern then repeats, later meetings occur at steps 32, 56, and so on.
