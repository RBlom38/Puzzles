## Question

Put eight points on a circle and draw a straight chord between every pair of points. Assume no three chords cross at the same interior point.

How many regions can the chords divide the circle into?

A tempting guess is that each new chord just adds one region. But a chord that crosses several existing chords can add more than one.

## Solution

Start with one region. Each chord adds one new region, plus one for every interior crossing it makes with earlier chords. There are $\binom{n}{2}$ chords. Assuming no three chords cross at one interior point, each selection of four circle points contributes exactly one crossing (the pair whose endpoints alternate). Thus

$R(n)=1+\binom{n}{2}+\binom{n}{4}.$

For eight points,

$R(8)=1+\binom{8}{2}+\binom{8}{4}=1+28+70=\mathbf{99}.$

## Python


```python
from itertools import combinations
from math import comb, pi

n = 8
chords = list(combinations(range(n), 2))

# Two chords cross inside the circle exactly when their endpoints alternate.
crossings = 0
crossing_pairs = []
for (a, b), (c, d) in combinations(chords, 2):
    if a < c < b < d or c < a < d < b:
        crossings += 1
        crossing_pairs.append(((a, b), (c, d)))

regions = 1 + len(chords) + crossings
formula_regions = 1 + comb(n, 2) + comb(n, 4)

print("Points:", n)
print("Chords:", len(chords))
print("Interior crossings:", crossings)
print("Regions by counting:", regions)
print("Regions by formula:", formula_regions)
print("\nExample crossing chords (their endpoints alternate around the circle):")
for first, second in crossing_pairs[:5]:
    print(f"  {first} crosses {second}")
```

```python
print("A crossing is determined entirely by endpoint order on the circle.")
print("For example, endpoints 0, 1, 2, 3 in that order give crossing chords (0, 2) and (1, 3).")
print("The code above checks every pair of chords using this alternating-endpoint rule.")
```

## Python interpretation

The code finds 28 chords and 70 interior crossings, then counts 99 regions. Its pairwise endpoint check and the formula agree. The assumption that no three chords meet at one interior point is essential.
