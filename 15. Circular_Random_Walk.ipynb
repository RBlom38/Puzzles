## Question

A token starts at position 0 on a circular track with 12 equally spaced positions. On every step, a fair coin decides whether it moves one position clockwise or one position counterclockwise.

The walk ends the first time the token reaches position 6, directly opposite its start.

How many steps do you expect this to take?

## Solution

Let $E_d$ be the expected number of steps remaining when the token is distance $d$ from the target along the shorter arc. The target is at distance 0, so $E_0=0$. For $1\le d<6$, a step moves one space closer or farther with equal probability:

$E_d=1+\frac{E_{d-1}+E_{d+1}}{2}.$

At the opposite point, distance 6, either direction goes to distance 5, so $E_6=1+E_5$. Solving gives $E_d=d(12-d)$, hence

$E_6=6(12-6)=\mathbf{36\ steps}.$

## Python


```python
import random

rng = random.Random()
track_size = 12
target = track_size // 2
games = 50_000
step_counts = []

for _ in range(games):
    position = 0
    steps = 0

    while position != target:
        direction = rng.choice((-1, 1))
        position = (position + direction) % track_size
        steps += 1

    step_counts.append(steps)

print(f"Simulated average over {games:,} walks: {sum(step_counts) / games:.2f} steps")
print("Exact expected time: 36 steps")
```

## Python interpretation

The simulation estimates the average from 50,000 walks; it should be close to the exact expectation of 36 steps. Individual walks can take much longer than the six-step shortest route.
