## Question

Forty-one people stand in a circle, numbered 1 through 41. Starting at person 1, count around the circle and eliminate every third person. Counting resumes at the next person still in the circle.

Which position is the last one remaining? Can you find it without simulating the whole circle by hand?

## Solution

Use the zero-based recurrence for the survivor's position among $n$ people:

$J(1)=0,\qquad J(n)=(J(n-1)+3)\bmod n.$

The $+3$ accounts for the shift in counting after each elimination. Applying the recurrence through $n=41$ gives zero-based position 30, or position 31 when numbered from 1.

## Python


```python
people = list(range(1, 42))
index = 0
eliminated = []
step = 3

while len(people) > 1:
    index = (index + step - 1) % len(people)
    eliminated.append(people.pop(index))

print("Last person remaining:", people[0])
print("First ten eliminated:", eliminated[:10])
print("Number eliminated:", len(eliminated))

# The classic recurrence uses zero-based positions.
survivor = 0
for size in range(1, 42):
    survivor = (survivor + step) % size

print("Recurrence check (converted to 1-based position):", survivor + 1)
```

## Python interpretation

The direct elimination simulation leaves person 31. The recurrence independently gives the same one-based position, checking the result by a second method.
