## Question

How many ways can eight queens be placed on a standard chessboard so that no two queens attack each other? Count rotations and reflections as different arrangements.


## Solution

There are **92** valid arrangements when rotations and reflections count as distinct. Place one queen in each row, and reject any square sharing a column or diagonal with a queen already placed. The recursive search explores every placement that remains valid and records a solution only after all eight rows are filled.

## Python

```python
board_size = 8
solutions = []

def place_queens(row, columns, descending_diagonals, ascending_diagonals, placement):
    if row == board_size:
        solutions.append(tuple(placement))
        return

    for column in range(board_size):
        descending_diagonal = row - column
        ascending_diagonal = row + column
        if (
            column not in columns
            and descending_diagonal not in descending_diagonals
            and ascending_diagonal not in ascending_diagonals
        ):
            columns.add(column)
            descending_diagonals.add(descending_diagonal)
            ascending_diagonals.add(ascending_diagonal)
            placement.append(column)

            place_queens(
                row + 1,
                columns,
                descending_diagonals,
                ascending_diagonals,
                placement,
            )

            placement.pop()
            columns.remove(column)
            descending_diagonals.remove(descending_diagonal)
            ascending_diagonals.remove(ascending_diagonal)

place_queens(0, set(), set(), set(), [])

print(f"Number of valid arrangements: {len(solutions)}")
print("One arrangement (column number for each row):", tuple(column + 1 for column in solutions[0]))
```

## Python interpretation

The backtracking search finds 92 arrangements. Each tuple lists the column of the queen in each row; rotations and reflections are not collapsed into a single arrangement.
