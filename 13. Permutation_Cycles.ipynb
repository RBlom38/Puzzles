## Question

Label eight cards 0 through 7. Shuffle them, then map each card to the position where it landed. This mapping is a permutation, and every permutation breaks into one or more cycles. For example, `0 → 3 → 0` is a cycle of length 2.

If all 8! shuffles are equally likely, how many cycles should you expect?

## Solution

For a uniformly random permutation of $n$ objects, the expected number of cycles is

$H_n=1+\frac12+\frac13+\cdots+\frac1n.$

Build the permutation one label at a time: label $k$ starts a new cycle with probability $1/k$. Adding these probabilities gives $H_n$. For eight cards,

$H_8=1+\frac12+\frac13+\cdots+\frac18\approx 2.717857,$

so the expected number is about 2.718 cycles.

### Python enumeration

```python
from collections import Counter
from itertools import permutations

n = 8
cycle_counts = Counter()

for permutation in permutations(range(n)):
    visited = set()
    cycles = 0

    for start in range(n):
        if start not in visited:
            cycles += 1
            position = start
            while position not in visited:
                visited.add(position)
                position = permutation[position]

    cycle_counts[cycles] += 1

total_shuffles = sum(cycle_counts.values())
average_cycles = sum(cycles * count for cycles, count in cycle_counts.items()) / total_shuffles
harmonic_number = sum(1 / k for k in range(1, n + 1))

print(f"Shuffles checked: {total_shuffles:,}")
print("Cycle-count distribution:")
for cycles in sorted(cycle_counts):
    print(f"  {cycles} cycles: {cycle_counts[cycles]:5,} shuffles")
print(f"\nEnumerated average: {average_cycles:.6f}")
print(f"Harmonic-number prediction: {harmonic_number:.6f}")
```

## Python interpretation

The enumeration checks all 40,320 shuffles and gives an average of **2.717857 cycles**, matching the harmonic-number result. The average is exact for these eight-card permutations, apart from floating-point rounding.
