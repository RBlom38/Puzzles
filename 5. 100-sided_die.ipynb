## Question

You roll a fair 100-sided die. You may stop after any roll and collect the face value, or pay $1 to discard it and roll again. What is the optimal stopping rule, and what is the expected net payout?

## Solution

Let $V$ be the expected net payout from a fresh roll when you follow the best strategy. If the current roll is $x$, stopping pays $x$. Rerolling costs $1$ and starts the same decision process over, so its expected net payout is $V-1$. Any fees paid on earlier rolls are already sunk and do not affect this choice. Thus, stop when $x\ge V-1$.

Write $c=V-1$ for the value of rerolling. If $86<c<87$, the rule is to reroll on 1–86 and stop on 87–100. Under this rule, a fresh roll has expected net payout

$V=\frac{86c+(87+88+\cdots+100)}{100}=\frac{86c+1309}{100}.$

Since $c=V-1$,

$100c=86c+1209,\qquad c=\frac{1209}{14}\approx 86.36.$

This confirms $86<c<87$, so the assumed rule is consistent and optimal: reroll on 1–86 and stop on 87–100. The expected net payout from a fresh roll is

$V=c+1=\frac{1223}{14}\approx \$87.36.$

Each reroll costs $1$; the initial roll is free.

## Python visualization


```python
import numpy as np
import matplotlib.pyplot as plt

thresholds = np.arange(1, 101)
p = (101 - thresholds) / 100
average_prize = (thresholds + 100) / 2
expected_rerolls = (1 - p) / p
net_value = average_prize - expected_rerolls

best_t = thresholds[np.argmax(net_value)]
print(best_t, net_value.max())

plt.plot(thresholds, net_value)
plt.axvline(best_t, color="red", label=f"Best threshold: {best_t}")
plt.xlabel("Stop if roll is at least")
plt.ylabel("Expected net payout ($)")
plt.title("Expected Value by Stopping Threshold")
plt.show()
```
