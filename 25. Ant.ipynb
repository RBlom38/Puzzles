## Question

An ant moves on an infinite grid of black and white squares. It starts on a white square facing north. At each step:

1. On a white square, turn right; on a black square, turn left.
2. Flip the square's color.
3. Move forward one square.

What does the ant's path look like after 12,000 steps? Does this simple rule keep wandering randomly, or does a larger pattern emerge?

## Solution

The ant follows a deterministic rule: on a white square it turns right, on a black square it turns left, then it flips that square and moves forward. The path initially looks irregular, but in the standard starting configuration it eventually settles into a repeating diagonal highway. After 12,000 steps, the ant is traveling along this repeated pattern rather than wandering randomly. This behavior is an observed outcome of iterating the rule, not a conclusion that follows from a short local calculation.

## Python


```python
steps = 12_000
black_squares = set()
x = y = 0
direction = 0  # north, east, south, west
directions = [(0, 1), (1, 0), (0, -1), (-1, 0)]

checkpoints = {100, 1_000, 10_000, steps}
positions = {}

for step in range(1, steps + 1):
    position = (x, y)
    if position in black_squares:
        direction = (direction - 1) % 4
        black_squares.remove(position)
    else:
        direction = (direction + 1) % 4
        black_squares.add(position)

    dx, dy = directions[direction]
    x += dx
    y += dy

    if step in checkpoints:
        positions[step] = (x, y, len(black_squares))

print("Checkpoint: (step, ant x, ant y, black squares)")
for step, state in positions.items():
    print(f"  {step:>5}: {state}")

radius = 10
print("\nLocal pattern around the ant (# black, . white, @ ant):")
for row_y in range(y + radius, y - radius - 1, -1):
    row = []
    for col_x in range(x - radius, x + radius + 1):
        if (col_x, row_y) == (x, y):
            row.append("@")
        elif (col_x, row_y) in black_squares:
            row.append("#")
        else:
            row.append(".")
    print("".join(row))
```

## Python interpretation

The simulation records the ant's position and visited black squares at selected checkpoints, then displays a local view of the grid. The full state is the set of black-square coordinates; the displayed window is only a portion of the board.
