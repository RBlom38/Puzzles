## Question

You may roll a fair six-sided die up to three times. After each roll, you can either keep the face value as your payout and stop, or reroll and give up the current value. If you reroll twice, you must keep the third roll.

What is the optimal strategy after each roll, and what is the expected payout if you play optimally?

## Solution

Work backward from the final roll. A fair six-sided die has an average roll of $3.50. After the second roll, keep 4, 5, or 6; reroll 1, 2, or 3 for one final chance.

The expected value after the second roll is

$\frac{3(3.50)+4+5+6}{6}=4.25.$

Before the second roll, compare the first roll with $4.25: keep 5 or 6; otherwise reroll. The optimal expected payout is

$\frac{4(4.25)+5+6}{6}=\frac{14}{3}\approx \$4.67.$

### Optimal strategy

- After roll 1: keep 5 or 6; reroll 1–4.
- After roll 2: keep 4, 5, or 6; reroll 1–3.
- After roll 3: keep the result.

## Python strategy check


```python
from itertools import product

faces = range(1, 7)
last_roll_value = sum(faces) / 6

best_value = 0
best_first_keep = None
best_second_keep = None

for first_keep in product([False, True], repeat=6):
    for second_keep in product([False, True], repeat=6):
        second_roll_value = 0
        for face in faces:
            if second_keep[face - 1]:
                second_roll_value += face
            else:
                second_roll_value += last_roll_value
        second_roll_value /= 6

        total_value = 0
        for face in faces:
            if first_keep[face - 1]:
                total_value += face
            else:
                total_value += second_roll_value
        total_value /= 6

        if total_value > best_value:
            best_value = total_value
            best_first_keep = first_keep
            best_second_keep = second_keep

print("Best expected payout:", best_value)
print("Keep after roll 1:", [face for face in faces if best_first_keep[face - 1]])
print("Keep after roll 2:", [face for face in faces if best_second_keep[face - 1]])
```

## Python simulation check


```python
import random

rng = random.Random()
games = 500_000
total = 0

for _ in range(games):
    roll = rng.randint(1, 6)
    if roll < 5:
        roll = rng.randint(1, 6)
        if roll < 4:
            roll = rng.randint(1, 6)
    total += roll

print(total / games)
```
