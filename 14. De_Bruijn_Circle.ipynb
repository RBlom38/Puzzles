## Question

A circular binary lock reads four consecutive bits at a time. You want to test every possible four-bit code—0000 through 1111—without repeating a code.

What is the shortest circular string that contains every four-bit pattern exactly once as a consecutive window?

The string wraps around at the end.

## Solution

There are $2^4=16$ different four-bit codes. A circular string of length 16 has exactly 16 starting positions for a four-bit window, so no shorter string can contain all 16 codes. The Python search finds a length-16 circle in which every code appears exactly once. Therefore, the shortest possible length is 16 bits.


## Python


```python
from itertools import product

order = 4
all_codes = {"".join(bits) for bits in product("01", repeat=order)}
length = len(all_codes)
answer = None

# Check every circular bit string of the minimum possible length.
for bits in product("01", repeat=length):
    windows = {
        "".join(bits[(start + offset) % length] for offset in range(order))
        for start in range(length)
    }
    if windows == all_codes:
        answer = "".join(bits)
        break

assert answer is not None
print("One shortest circular string:", answer)
print("Length:", len(answer))
windows = [
    "".join(answer[(start + offset) % length] for offset in range(order))
    for start in range(length)
]
print("Distinct four-bit windows:", len(set(windows)))
print("All 16 codes appear exactly once:", set(windows) == all_codes)
```

## Python interpretation

The search finds `0000100110101111`, with 16 distinct four-bit windows. The code verifies that every possible code appears exactly once, attaining the lower bound.
