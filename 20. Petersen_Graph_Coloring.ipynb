## Question

The Petersen graph has ten vertices. Draw an outer pentagon, connect each outer vertex to one inner vertex, then connect the inner vertices in a five-point star. Adjacent vertices must have different colors.

What is the fewest number of colors needed?

## Solution

At least three colors are needed: the outer five vertices form a 5-cycle, which cannot be colored with only two colors because alternating colors around an odd cycle makes the last vertex conflict with the first.

Three colors are sufficient. For the graph labeling in the code, one valid coloring is color 1 on vertices $\{0,2,6\}$, color 2 on $\{1,3,5,9\}$, and color 3 on $\{4,7,8\}$. No edge has both endpoints in the same group. Therefore, the Petersen graph's chromatic number is 3.

## Python


```python
from itertools import combinations

vertices = range(10)
edges = set()

def add_edge(a, b):
    edges.add(tuple(sorted((a, b))))

# Outer pentagon and five spokes.
for i in range(5):
    add_edge(i, (i + 1) % 5)
    add_edge(i, i + 5)

# Inner five-point star.
for i in range(5):
    add_edge(i + 5, ((i + 2) % 5) + 5)

neighbors = {v: set() for v in vertices}
for a, b in edges:
    neighbors[a].add(b)
    neighbors[b].add(a)

order = sorted(vertices, key=lambda v: -len(neighbors[v]))

def find_coloring(number_of_colors):
    colors = {}

    def assign_next(index):
        if index == len(order):
            return colors.copy()

        vertex = order[index]
        for color in range(number_of_colors):
            if all(colors.get(other) != color for other in neighbors[vertex]):
                colors[vertex] = color
                answer = assign_next(index + 1)
                if answer is not None:
                    return answer
                del colors[vertex]
        return None

    return assign_next(0)

for number_of_colors in range(1, 5):
    coloring = find_coloring(number_of_colors)
    print(f"{number_of_colors} color(s):", "possible" if coloring else "impossible")
    if coloring:
        break

print("Color assignment:", coloring)
assert all(coloring[a] != coloring[b] for a, b in edges)
```

```python
for color in range(max(coloring.values()) + 1):
    group = [vertex for vertex in vertices if coloring[vertex] == color]
    print(f"Color {color + 1}:", group)

print("Edges:", sorted(edges))
print("Every edge joins vertices with different colors:",
      all(coloring[a] != coloring[b] for a, b in edges))
```

## Python interpretation

The backtracking search rejects one- and two-color assignments and finds a valid three-coloring. The final check confirms that every edge joins differently colored vertices, establishing that three colors suffice; the odd outer pentagon establishes that fewer do not.
