## Question

You start with $3. Each round, a fair coin changes your balance by $1: heads adds $1, tails subtracts $1. You stop when your balance reaches $0 or $10.

1. What is the probability you reach $10 before going broke?
2. How many rounds do you expect the game to last?

Find the exact values, then use Python simulation to check them.

## Solution

Let $p_i$ be the probability of reaching $10$ before $0$ when starting with $i$ dollars. On the next fair-coin toss, the balance moves to $i-1$ or $i+1$, so

$p_i=\frac{p_{i-1}+p_{i+1}}{2},\qquad p_0=0,\quad p_{10}=1.$

The solution is $p_i=i/10$, so from $3$ dollars the chance of reaching $10$ is **$3/10$**.

Let $E_i$ be the expected number of rounds remaining from balance $i$. One round is used, then the balance changes up or down with equal probability:

$E_i=1+\frac{E_{i-1}+E_{i+1}}{2},\qquad E_0=E_{10}=0.$

This recurrence has solution $E_i=i(10-i)$. Starting at $3$, the expected game length is **$21$ rounds**.

## Python simulation


```python
import random

starting_balance = 3
target_balance = 10
trials = 100_000
rng = random.Random(42)

wins = 0
total_rounds = 0

for _ in range(trials):
    balance = starting_balance
    rounds = 0

    while 0 < balance < target_balance:
        balance += 1 if rng.randrange(2) else -1
        rounds += 1

    wins += balance == target_balance
    total_rounds += rounds

print(f"Estimated chance of reaching $10: {wins / trials:.4f} (exact: 0.3000)")
print(f"Estimated game length: {total_rounds / trials:.2f} rounds (exact: 21)")
```
