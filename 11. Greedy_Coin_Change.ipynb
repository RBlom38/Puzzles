## Question

A vending machine has unlimited coins worth 1, 3, and 4 units. To return 6 units of change, a greedy cashier repeatedly gives the largest coin that does not exceed the amount still owed.

The greedy method gives 4 + 1 + 1, using three coins. Is that the fewest possible? Can you find a method that guarantees the fewest coins for any amount?

## Solution

The greedy method uses $4+1+1$, or three coins. But $3+3=6$, so two coins are enough; one coin cannot make 6. This means the minimum is two coins.

For any amount $v$, let $best[v]$ be the fewest coins needed to make it, with $best[0]=0$. Consider every coin $c\le v$ as the last coin used. Then

$best[v]=1+\min_{c\le v} best[v-c].$

Computing amounts from small to large ensures each smaller answer is already known. Recording which coin achieves the minimum lets us reconstruct the coin combination. Unlike greedy selection, this checks every possible last-coin choice, so it finds an optimal solution.

## Python


```python
coins = [1, 3, 4]
amount = 6

# Greedy: repeatedly take the largest coin that fits
remaining = amount
greedy_coins = []
for coin in sorted(coins, reverse=True):
    while coin <= remaining:
        greedy_coins.append(coin)
        remaining -= coin

# Dynamic programming: best[a] is the fewest coins needed to make a.
best = [0] + [float("inf")] * amount
choice = [None] * (amount + 1)
for value in range(1, amount + 1):
    for coin in coins:
        if coin <= value and best[value - coin] + 1 < best[value]:
            best[value] = best[value - coin] + 1
            choice[value] = coin

optimal_coins = []
value = amount
while value:
    coin = choice[value]
    optimal_coins.append(coin)
    value -= coin

print("Greedy change:", greedy_coins, "-", len(greedy_coins), "coins")
print("Fewest-coin change:", optimal_coins, "-", int(best[amount]), "coins")
```

## Python interpretation

The output confirms that greedy change uses three coins ($4+1+1$), while dynamic programming finds the two-coin optimum ($3+3$).
