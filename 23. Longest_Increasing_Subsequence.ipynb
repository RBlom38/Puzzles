## Question

Take a random shuffle of the numbers 1 through 8. A rising subsequence is a selection of entries, kept in their original order, whose values increase. The selected entries do not need to be next to each other.

For example, in `3, 1, 5, 2, 6`, one rising subsequence is `1, 2, 6`.

Across all 8! shuffles, what is the average length of the longest rising subsequence?

## Solution

For each permutation, the algorithm computes the longest increasing subsequence (LIS) length. As it scans the sequence, `tails[k]` stores the smallest possible ending value for an increasing subsequence of length $k+1$. Keeping the smallest ending value preserves the greatest opportunity to extend a subsequence later; the number of entries in `tails` is the LIS length.

Enumerating all $8!=40{,}320$ permutations and averaging their LIS lengths gives the exact mean

$\frac{7601}{2016}\approx\mathbf{3.770337}.$

## Python


```python
from bisect import bisect_left
from collections import Counter
from fractions import Fraction
from itertools import permutations

def longest_increasing_length(sequence):
    # tails[k] is the smallest possible last value of an increasing
    # subsequence of length k + 1 seen so far.
    tails = []
    for value in sequence:
        position = bisect_left(tails, value)
        if position == len(tails):
            tails.append(value)
        else:
            tails[position] = value
    return len(tails)

length_counts = Counter()
for shuffle in permutations(range(1, 9)):
    length_counts[longest_increasing_length(shuffle)] += 1

total = sum(length_counts.values())
average = Fraction(sum(length * count for length, count in length_counts.items()), total)

print("Shuffles checked:", total)
print("Longest-subsequence length distribution:")
for length in sorted(length_counts):
    print(f"  length {length}: {length_counts[length]:5,} shuffles")
print(f"\nExact average: {average} = {float(average):.6f}")
```

## Python interpretation

The enumeration counts all 40,320 shuffles and reports an exact average of $7601/2016$, or about **3.770337**. Because every shuffle is included, this is not a sampling estimate.
